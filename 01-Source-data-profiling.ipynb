Create the data storage

```python
%sql
CREATE CATALOG IF NOT EXISTS workspace;
CREATE SCHEMA IF NOT EXISTS workspace.skypoints;
CREATE VOLUME IF NOT EXISTS workspace.skypoints.raw_files;
```

Define input paths

```python
IND_PATH = "/Volumes/workspace/skypoints/raw_files/IND.csv"
USA_PATH = "/Volumes/workspace/skypoints/raw_files/USA.csv"
AUS_PATH = "/Volumes/workspace/skypoints/raw_files/AUS.csv"
```

Read the three CSV files

```python
from pyspark.sql import functions as F
```

```python
ind_df = (
    spark.read
    .option("header",True)
    .option("inferSchema", False)
    .csv(IND_PATH)
)

usa_df = (
    spark.read
    .option("header",True)
    .option("inferSchema", False)
    .csv(USA_PATH)
)

aus_df = (
    spark.read
    .option("header",True)
    .option("inferSchema", False)
    .csv(AUS_PATH)
)
```

```python
display(ind_df)
display(usa_df)
display(aus_df)
```
